## Inspect the resulting image
This has photometric conversion metadata and flat-fielding, but no source extraction or fitted astrometric alignment yet.


```python
import roman_datamodels as rdm
import matplotlib.pyplot as plt
from astropy.visualization import simple_norm
with rdm.open(str(current)) as model:
    print(model.meta.cal_step)
    print(model.meta.photometry)
    plt.imshow(model.data, origin='lower', cmap='gray', norm=simple_norm(model.data, 'asinh', percent=99))
    plt.show()
```
